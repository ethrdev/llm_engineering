# Kleinigkeiten, die im Kurs auftreten
Dies ist zwar keine spezielle Funktion von Jupyter, aber ein hilfreiches Package, das häufig in Jupyter Lab verwendet wird.
Das Package `tqdm` zeigt einen ansprechenden Fortschrittsbalken, wenn du ein iterierbares Objekt damit umgibst.

```python
# Beispielcode ohne Fortschrittsbalken
# Dieser Code benötigt etwa 10 Sekunden, während du beobachten kannst, was passiert.
import time

spams = ["spam"] * 1000

for spam in spams:
    time.sleep(0.01)
```

```python
# Beispielcode mit Fortschrittsbalken
# Hier wird das Package 'tqdm' verwendet, um den Fortschritt der Schleife anzuzeigen.
import time
from tqdm import tqdm

spams = ["spam"] * 1000

for spam in tqdm(spams):
    time.sleep(0.01)
```

```python
# Ausgabe von Markdown direkt im Code
# Mithilfe von 'IPython.display' kannst du Markdown-Text formatiert ausgeben.
from IPython.display import Markdown, display

display(Markdown("# This is a big heading!\n\n- And this is a bullet-point\n- So is this\n- Me, too!"))
```
