## Weiter geht's mit unserem ersten Projekt

```python
# HTTP-Header für Web Scraping-Anfragen (um von Websites nicht als Bot blockiert zu werden)
headers = {
 "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36"
}

class Website:
    """
    Klasse zur Repräsentation einer Webseite.
    Beim Erstellen eines Objekts wird die URL abgerufen und der HTML-Inhalt automatisch geparst.
    """
    
    def __init__(self, url):
        """
        Initialisiert ein Website-Objekt anhand einer URL.
        Ruft den HTML-Inhalt ab und parst ihn mittels BeautifulSoup.
        """
        self.url = url
        response = requests.get(url, headers=headers)
        soup = BeautifulSoup(response.content, 'html.parser')

        # Extrahiere den Titel der Seite (falls vorhanden)
        self.title = soup.title.string if soup.title else "No title found"

        # Entferne unerwünschte Elemente (z. B. script, style, img, input) aus dem Body
        for irrelevant in soup.body(["script", "style", "img", "input"]):
            irrelevant.decompose()

        # Extrahiere den reinen Textinhalt des Body-Bereichs
        self.text = soup.body.get_text(separator="\n", strip=True)
```

```python
# Teste die Website-Klasse.
# Beispiel: Erstelle ein Website-Objekt f�r "https://edwarddonner.com"

# Ausgabe des Titels und des extrahierten Textes der Webseite
ed = Website("https://edwarddonner.com")
print(ed.title)
print(ed.text)
```

## Arten von Prompts

Möglicherweise kennst Du das bereits – falls nicht, wirst Du Dich intensiv damit beschäftigen!

Modelle wie GPT4o erwarten Anweisungen in einem bestimmten Format:

- **Ein system prompt:** Definiert, welche Aufgabe das Modell ausführen soll und in welchem Ton.
- **Ein user prompt:** Stellt den Gesprächseinstieg dar, auf den das Modell reagieren soll.

```python
# Definiere unseren system prompt \u2013 Du kannst sp�ter experimentieren, z. B. indem Du den letzten Satz zu "Respond in markdown in Spanish" �nderst.

system_prompt = "You are an assistant that analyzes the contents of a website \
and provides a short summary written in german language, ignoring text that might be navigation related. \
Respond in markdown."
```

```python
# Funktion zur Erstellung eines User Prompts, der eine Zusammenfassung einer Webseite anfordert:
def user_prompt_for(website):
    user_prompt = f"You are looking at a website titled {website.title}"
    user_prompt += "\nThe contents of this website is as follows; \
please provide a short summary of this website in markdown written in german language. \
If it includes news or announcements, then summarize these too.\n\n"
    user_prompt += website.text
    return user_prompt
```

```python
# Ausgabe des generierten User Prompts für das Website-Objekt 'ed'
print(user_prompt_for(ed))
```

## Nachrichten

Die OpenAI API erwartet Nachrichten in einem bestimmten Format \u2013 dieses Schema wird auch von den meisten anderen LLMs verwendet.  
Beispielsweise:

[
    {"role": "system", "content": "system message goes here"},
    {"role": "user", "content": "user message goes here"}
]

Dies ist eine Python-Liste, die Dictionaries enth�lt.  
- Das erste Dictionary hat den Schl�ssel "role" mit dem Wert "system" und den Schl�ssel "content" mit dem Text "system message goes here".  
- Das zweite Dictionary hat den Schl�ssel "role" mit dem Wert "user" und den Schl�ssel "content" mit dem Text "user message goes here".

```python
# Beispiel: Erstellung einer Nachrichtenliste (messages) im von OpenAI erwarteten Format
messages = [
    {"role": "system", "content": "You are a snarky assistant"},
    {"role": "user", "content": "What is 2 + 2?"}
]
```

```python
# Vorschau: Aufruf der OpenAI API mit System- und User-Nachrichten
response = openai.chat.completions.create(model="gpt-4o-mini", messages=messages)
print(response.choices[0].message.content)
```

## Und nun: Erstellen hilfreicher Nachrichten für GPT-4o-mini mithilfe einer Funktion

```python
# Diese Funktion erzeugt exakt das oben beschriebene Format f�r Nachrichten
def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_for(website)}
    ]
```

```python
# Teste diese Funktion und probiere sie anschließend mit weiteren Webseiten aus
messages_for(ed)
```

Jetzt fassen wir alles zusammen \u2013 die OpenAI API ist sehr einfach!

```python
# Rufe die OpenAI API auf – so wirst Du mit ihr vertraut!
# Wir erstellen eine Funktion namens summarize, die eine URL als Parameter entgegennimmt.
def summarize(url):
    # Erzeuge ein Website-Objekt für die angegebene URL
    website = Website(url)
    # Rufe die Completion-API von OpenAI auf
    response = openai.chat.completions.create(
        model = "gpt-4o-mini",
        # Übergib die zuvor erstellten Nachrichten
        messages = messages_for(website)
    )
    # Gib die Antwort des API-Aufrufs zurück
    return response.choices[0].message.content
```

```python
# Teste die summarize-Funktion mit der URL "https://cnn.com"
summarize("https://cnn.com")
```

```python
# Funktion zur ansprechenden Darstellung der Zusammenfassung in Jupyter mittels Markdown
def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
# Zeige die Zusammenfassung f�r "https://edwarddonner.com" an
display_summary("https://edwarddonner.com")
```
